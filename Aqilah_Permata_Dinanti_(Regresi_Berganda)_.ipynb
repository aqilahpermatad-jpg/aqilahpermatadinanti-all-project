# Analisis Kemiskinan Dengan Regresi Berganda Kota Tanjung Balai

---

`indikator variabel`

*   Y (Variabel Dependen) = KEMISKINAN
*   X1 (Variabel Independen) = DAU_Earmark (Dana Alokasi Umum)
*   X2 (Variabel Indpenden) = DAU_Grant (Dana Alokasi Umum)








```python
import pandas as pd

# 1. Load dataset utama
df_full = pd.read_csv('dataset_fiskal_97_hipoFull (3).csv')

# 2. Filter khusus Kota Tanjung Balai
df_tb = df_full[df_full['Kota'] == 'Kota Tanjung Balai'].copy()

# 3. Membuat DataFrame baru dengan indikator yang dipilih
df_regresi_tb = df_tb[['Tahun', 'DAU_Earmark', 'DAU_Grant', 'KEMISKINAN']].reset_index(drop=True)
df_regresi_tb.columns = ['Tahun', 'DAU_Earmark_X1', 'DAU_Grant_X2', 'Kemiskinan_Y']

print("DataFrame Baru - Kota Tanjung Balai:")
print(df_regresi_tb)
```

```python
import numpy as np
import statsmodels.api as sm
import plotly.graph_objects as go

x1 = df_regresi_tb['DAU_Earmark_X1']
x2 = df_regresi_tb['DAU_Grant_X2']
y  = df_regresi_tb['Kemiskinan_Y']

X = df_regresi_tb[['DAU_Earmark_X1', 'DAU_Grant_X2']]

# 1. Menambahkan Konstanta Intercept
X_model = sm.add_constant(X)

# 2. Estimasi Model OLS
model = sm.OLS(y, X_model)
results = model.fit()

# Menampilkan Ringkasan Hasil Regresi
print("=" * 70)
print("HASIL REGRESI LINEAR BERGANDA (KOTA TANJUNG BALAI)")
print("=" * 70)
print(results.summary())

# 3. Mengambil Parameter
intercept = results.params["const"]
coef_x1 = results.params["DAU_Earmark_X1"]
coef_x2 = results.params["DAU_Grant_X2"]
r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

# 4. Membuat Persamaan Regresi
persamaan = (
    f"Y = {intercept:.3f}"
    f" + ({coef_x1:.3f}) X1"
    f" + ({coef_x2:.3f}) X2"
)
print("\n" + "=" * 70)
print("PERSAMAAN REGRESI:")
print(persamaan)
print(f"R² = {r_squared:.4f} | Adjusted R² = {adjusted_r_squared:.4f}")
print("=" * 70)

# 5. Membuat Grid untuk Bidang Regresi 3D
grid_x1 = np.linspace(x1.min(), x1.max(), 40)
grid_x2 = np.linspace(x2.min(), x2.max(), 40)
X1_grid, X2_grid = np.meshgrid(grid_x1, grid_x2)
Y_grid = intercept + (coef_x1 * X1_grid) + (coef_x2 * X2_grid)

# 6. Membuat Figure 3D Plotly
fig = go.Figure()

# Plot Data Aktual (Titik Merah)
fig.add_trace(
    go.Scatter3d(
        x=x1,
        y=x2,
        z=y,
        mode="markers",
        marker=dict(size=7, color="red", opacity=0.9),
        name="Data Aktual",
        text=[f"Tahun: {t}" for t in df_regresi_tb['Tahun']],
        hovertemplate=(
            "<b>%{text}</b><br>"
            "DAU Earmark (X1): %{x}<br>"
            "DAU Grant (X2): %{y}<br>"
            "Kemiskinan (Y): %{z}"
            "<extra></extra>"
        )
    )
)

# Plot Bidang Regresi (Surface)
fig.add_trace(
    go.Surface(
        x=X1_grid,
        y=X2_grid,
        z=Y_grid,
        opacity=0.60,
        colorscale="Viridis",
        name="Bidang Regresi",
        hovertemplate=(
            "DAU Earmark (X1): %{x:.2f}<br>"
            "DAU Grant (X2): %{y:.2f}<br>"
            "Prediksi Kemiskinan (Y): %{z:.2f}"
            "<extra></extra>"
        )
    )
)

# Pengaturan Layout Grafik 3D
fig.update_layout(
    title={
        "text": (
            "REGRESI LINEAR BERGANDA 3D - KOTA TANJUNG BALAI<br>"
            f"<sup>{persamaan} | R² = {r_squared:.3f}</sup>"
        ),
        "x": 0.5
    },
    scene={
        "xaxis": {"title": "DAU Earmark (X1)"},
        "yaxis": {"title": "DAU Grant (X2)"},
        "zaxis": {"title": "Kemiskinan (Y)"}
    },
    width=900,
    height=700,
    margin={"l": 0, "r": 0, "b": 0, "t": 80}
)

fig.show()
```
